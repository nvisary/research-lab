# 05 — seeing the book (the liquidity heatmap)

The problem: microstructure is too much data per second to read as a table. So
we render it the way a trader *sees* order flow — a **liquidity heatmap**:

- x = time, y = price, colour = resting size in the book at that price.
- **bids (support) below in blue, asks (resistance) above in warm** — bright
  horizontal bands are walls; watch them appear (refill) or vanish (pull/spoof).
- white line = mid price weaving through the book.
- **trade bubbles**: green = taker buy, red = taker sell, size ∝ amount.
- lower panels: **CVD** (net aggression) and **force** (taker vol/sec, buy vs sell).

One call: `micro_view(session, sym, t0=None, t1=None)` (in `_lab.py`). This is a
*replay* of a recording — we can zoom and re-look, and both of us see the same
PNG. Goal: look, find relationships, form hypotheses (then measure them).
